# **Q2). Plot MSE versus order m, for m = 1, 2, 3, 4, 5, 6, 7, 8, 9 respectively. Identify the best choice of m. Discuss how m impacts the performance in terms model accuracy and overfitting.**

```python
# Compute MSE for polynomial orders 1 to 9
m_values = list(range(1, 10))
mse_values = []

for m in m_values:
    coefficients = np.polyfit(x.flatten(), y.flatten(), m)
    y_pred = np.polyval(coefficients, x.flatten())
    mse = np.mean((y.flatten() - y_pred) ** 2)
    mse_values.append(mse)

plt.plot(m_values, mse_values, 'bo-', linewidth=2, markersize=8)
plt.xlabel('Polynomial Order (m)', fontsize=12)
plt.ylabel('Mean Squared Error (MSE)', fontsize=12)
plt.title('MSE vs. Polynomial Order', fontsize=14)
plt.grid(True)
plt.show()

best_m = m_values[np.argmin(mse_values)]
print(f"The best polynomial order is m = {best_m}")
```

Low m (m=1,2): The polynomial is too simple to capture the underlying trend in the data. This results in underfitting, where the model has high bias and poor accuracy on both the training and test data.

Moderate m (m=3,4,5): The polynomial is flexible enough to capture the underlying trend without fitting the noise. This is the optimal range where the model achieves good accuracy and generalizes well to unseen data.

High m (m=6,7,8,9): The polynomial becomes overly complex and starts fitting the noise in the training data. This results in overfitting, where the model has high variance and performs well on the training data but poorly on unseen data.
